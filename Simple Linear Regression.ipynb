```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import pandas as pd

# Sample dataset: CGPA vs Package
data = {
    "CGPA": [6.0, 6.5, 7.0, 7.5, 8.0, 8.5, 9.0, 9.1, 9.2, 9.3, 9.4, 9.5, 9.6, 9.7, 9.8, 9.9, 10.0],
    "Package": [3.0, 3.5, 4.0, 5.0, 6.0, 7.0, 8.5, 9.0, 9.2, 9.5, 10.0, 11.0, 12.0, 12.5, 13.0, 14.0, 15.0]
}

# Create a DataFrame
df = pd.DataFrame(data)

# Display the DataFrame
print(df)
```

```python
x=df[["CGPA"]]
y=df["Package"]
x.head(3),y.head(3)
```

```python
plt.figure(figsize=(4,4))
plt.scatter(x="CGPA",y="Package",data=df)
plt.show()
```

```python
#It follows linear regression and there is only one feature input.So,we use simple linear equation
```

```python
#Now we do train and test splitting
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=60)
```

```python
x_test
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr=LinearRegression()
lr.fit(x_train,y_train)
```

```python
lr.predict([[6.0]])
```

```python
lr.predict([[9.0]])
```

```python
lr.score(x_test,y_test)*100
```
