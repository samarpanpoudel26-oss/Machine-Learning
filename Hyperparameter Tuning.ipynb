```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset=pd.read_csv("Salary_Data.csv")
dataset.head()
```

```python
dataset.isnull().sum()
```

```python
x=dataset[["YearsExperience"]]
y=dataset[["Salary"]]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
model=DecisionTreeRegressor()
model.get_params()
```

```python
dt=DecisionTreeRegressor(criterion="absolute_error",random_state=13,splitter="random")
dt.fit(x_train,y_train)
```

```python
dt.score(x_test,y_test)*100
```

```python
dt.score(x_train,y_train)*100
```

Hyper Parameter Tuning

```python
from sklearn.model_selection import GridSearchCV
```

```python
df={"criterion" : ["squared_error", "friedman_mse", "absolute_error","poisson"],
   "splitter":["best", "random"],
    "random_state":[i for i in range(2,50)]
   }
```

```python
gd=GridSearchCV(DecisionTreeRegressor(),param_grid=df)
gd.fit(x_train,y_train)
```
